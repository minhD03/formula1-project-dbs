# Set up key in Databricks
- The Databricks access connector is created alongside the databricks account.
- Add the Access connector to the storage account in IAM => role: Storage Blob Data Contributor.
- In Databricks: Catalog => External Data => Credentials => Create credentials => Azure Managed Identity => Point to Access Connector.
- Go to Catalog => Create External Locations => Give Storage Url, Storage Credential and Access Connector Resource ID.
